```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
income = pd.read_csv('/content/drive/MyDrive/AI ML/income_dataset.csv')
```

```python
income.head()
```

```python
income
```

```python
income.info()
```

```python
income.describe()
```

```python
income.isnull().sum()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=income, x='Age', y='Income')

plt.title('Age vs Income')
plt.xlabel('Age')
plt.ylabel('Income')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=income, x='Education_Years', y='Income')

plt.title('Education vs Income')
plt.xlabel('Education Years')
plt.ylabel('Income')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=income, x='Experience_Years', y='Income')

plt.title('Experience vs Income')
plt.xlabel('Experience Years')
plt.ylabel('Income')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(income['Income'], kde=True)

plt.title('Income Distribution')
plt.xlabel('Income')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=income['Income'])

plt.title('Income Boxplot')
plt.ylabel('Income')

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    income.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap - Income Dataset')

plt.show()
```

```python
advertising = pd.read_csv('/content/drive/MyDrive/AI ML/Advertising.xls')
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=advertising, x='TV', y='Sales')

plt.title('TV Advertising vs Sales')
plt.xlabel('TV Advertising')
plt.ylabel('Sales')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=advertising, x='Radio', y='Sales')

plt.title('Radio Advertising vs Sales')
plt.xlabel('Radio Advertising')
plt.ylabel('Sales')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=advertising, x='Newspaper', y='Sales')

plt.title('Newspaper Advertising vs Sales')
plt.xlabel('Newspaper Advertising')
plt.ylabel('Sales')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(advertising['Sales'], kde=True)

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=advertising[['TV', 'Radio', 'Newspaper', 'Sales']]
)

plt.title('Advertising Data Distribution')

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    advertising.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap - Advertising Dataset')

plt.show()
```
